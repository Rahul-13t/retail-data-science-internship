```python
print("===== DATASET 1: ONLINE RETAIL =====")
print("Shape:", df.shape)
display(df.head())
df.info()

print("\nMissing Values:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nNumerical Summary:")
display(df.describe())

print("\n===== DATASET 2 =====")
print("Shape:", df2.shape)
display(df2.head())
df2.info()

print("\nMissing Values:")
display(df2.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate Rows:", df2.duplicated().sum())

print("\nNumerical Summary:")
display(df2.describe())
```

```python
df_clean = df.copy()

before_rows = len(df_clean)

df_clean["Quantity"] = pd.to_numeric(df_clean["Quantity"], errors="coerce")
df_clean["UnitPrice"] = pd.to_numeric(df_clean["UnitPrice"], errors="coerce")
df_clean["InvoiceDate"] = pd.to_datetime(df_clean["InvoiceDate"], errors="coerce")

df_clean = df_clean.dropna(subset=["InvoiceNo", "StockCode", "InvoiceDate", "Quantity", "UnitPrice"])
df_clean = df_clean.drop_duplicates()

df_clean = df_clean[
    (df_clean["Quantity"] > 0) &
    (df_clean["UnitPrice"] > 0)
]

df_clean["TotalSales"] = df_clean["Quantity"] * df_clean["UnitPrice"]

after_rows = len(df_clean)

print("Rows before cleaning:", before_rows)
print("Rows after cleaning:", after_rows)
print("Rows removed:", before_rows - after_rows)
print("Duplicates remaining:", df_clean.duplicated().sum())
print("Missing values remaining:", df_clean.isnull().sum().sum())

display(df_clean.head())
```

```python
print("Missing values after cleaning:")
display(df_clean.isnull().sum().to_frame("Missing Values"))

print("\nAbnormal values before/after cleaning:")
print("Quantity <= 0:", (df_clean["Quantity"] <= 0).sum())
print("UnitPrice <= 0:", (df_clean["UnitPrice"] <= 0).sum())
print("TotalSales <= 0:", (df_clean["TotalSales"] <= 0).sum())

print("\nFinal statistics:")
display(df_clean[["Quantity", "UnitPrice", "TotalSales"]].describe())
```

```python
Q1 = df_clean["TotalSales"].quantile(0.25)
Q3 = df_clean["TotalSales"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = ((df_clean["TotalSales"] < lower) | (df_clean["TotalSales"] > upper)).sum()

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
print("Number of outliers:", outliers)

df_clean["TotalSales_Capped"] = df_clean["TotalSales"].clip(lower, upper)

print("\nOutlier handling completed using IQR capping.")
```

```python
print("Dataset 2 numerical columns:")
print(df2.select_dtypes(include=np.number).columns.tolist())

print("\nDataset 2 missing values:")
display(df2.isnull().sum().to_frame("Missing Values"))

print("\nDataset 2 duplicate rows:", df2.duplicated().sum())
```

```python
df2_clean = df2.copy()

df2_clean = df2_clean.drop_duplicates()

num_cols = df2_clean.select_dtypes(include=np.number).columns

for col in num_cols:
    df2_clean[col] = pd.to_numeric(df2_clean[col], errors="coerce")

df2_clean[num_cols] = df2_clean[num_cols].fillna(
    df2_clean[num_cols].median()
)

print("Dataset 2 after cleaning:")
print("Shape:", df2_clean.shape)
print("Duplicates:", df2_clean.duplicated().sum())
print("Missing numerical values:", df2_clean[num_cols].isnull().sum().sum())

display(df2_clean.head())
```

```python
print("========== FINAL DATA VALIDATION ==========")

checks = {
    "Rows": len(df_clean),
    "Columns": len(df_clean.columns),
    "Duplicate Rows": df_clean.duplicated().sum(),
    "Missing Values": df_clean.isnull().sum().sum(),
    "Invalid Quantity": (df_clean["Quantity"] <= 0).sum(),
    "Invalid UnitPrice": (df_clean["UnitPrice"] <= 0).sum(),
    "Invalid TotalSales": (df_clean["TotalSales"] <= 0).sum()
}

for key, value in checks.items():
    print(f"{key}: {value}")
```
